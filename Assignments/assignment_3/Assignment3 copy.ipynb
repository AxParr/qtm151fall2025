- Use element-by-element vector operations to compute a new array `temps_f` of length 5, which will convert the temperatures from Celsius to Fahrenheit: <br><br>
`temps_f` $= \begin{pmatrix} f_1 \\[8pt] f_2 \\[8pt] f_3 \\[8pt] f_4 \\[8pt] f_5 \end{pmatrix} = \begin{pmatrix} c_1 \times \tfrac{9}{5} +32 \\[8pt] c_2 \times \tfrac{9}{5} +32 \\[8pt] c_3 \times \tfrac{9}{5} + 32 \\[8pt] c_4 \times \tfrac{9}{5} + 32 \\[8pt] c_5 \times \tfrac{9}{5} + 32\end{pmatrix} $ <br><br>

```python
# Defining array of temp_c
temps_c = np.array([0, 10, 20, 30, 40])
# Element-by-element vector operations for computation 
temps_f = temps_c * 9/5 + 32

# Printing temps_f
print(temps_f)

# Computing the average of the Fahrenheit temperatures
print("Average temperature in Fahrenheit:")
print(temps_f.mean())
```

(c) Plot a Normal Random Variable
<font size = "3">

To get full points:

- Set the seed using `np.random.seed(1234)`
- Make the random variable with the following parameters:
    -  mean = 7
    - standard deviation = 5
    - size = 1500

- Make a histogram
- Create a title
- Label the axes

```python
# Setting the seed
np.random.seed(1234)
sample = np.random.normal(loc = 7, scale= 5, size=1500)

# Histogram
plt.hist(x = sample, color = 'pink')
plt.title("Histogram of Random Variable")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show
```

(e) Classifying a Variable

<font size = "3">

- Below, the variable `pizza_rating` is assigned a random value between 0 and 10
- Write an if/elif/else statement that does the following
  - `print("Best Pizza Ever")` if the rating is **greater than or equal to 9**.  
  - `print("Pretty Good")` if the rating is **greater than or equal to 7 but less than 9**.  
  - `print("Meh")` if the rating is **greater than or equal to 4 but less than 7**.  
  - `print("Never Again")` otherwise.

- To get full points you must do all of the above tasks 

```python
pizza_rating = np.random.uniform(low = 0, high = 10)
print(pizza_rating)

# Statements
if pizza_rating >= 9:
    print("Best Pizza Ever")
elif pizza_rating >= 7 and pizza_rating < 9:
    print("Pretty Good")
elif pizza_rating >= 4 and pizza_rating < 7:
    print("Meh")
else:
    print("Never Again")
```
